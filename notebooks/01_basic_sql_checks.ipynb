```python
import sqlite3
import pandas as pd

conn=sqlite3.connect("../data/analytics.db")
```

```python
tables=pd.read_sql("SELECT name FROM sqlite_master WHERE type='table';", conn)
print(tables)
```

```python
for t in tables['name']:
    count=pd.read_sql(f"SELECT COUNT(*) as rows FROM {t};", conn)
    print(f"Table: {t}, Count: {count.iloc[0]['rows']}")
```

```python
pd.read_sql("SELECT * FROM products LIMIT 5;", conn)
```

```python
pd.read_sql("SELECT * FROM transaction_items LIMIT 5;", conn)
```

```python
pd.read_sql("SELECT * FROM products p join transaction_items ti on p.product_id=ti.product_id LIMIT 5;", conn)
```

```python
# Sum of prices in transaction_items
pd.read_sql("SELECT sum(price) FROM transaction_items group by transaction_id;", conn)
```

```python
# Max of prices in transaction_items
pd.read_sql("SELECT *,max(prices) FROM (SELECT *,sum(price) as prices FROM transaction_items group by transaction_id) as subquery;", conn)
```
